## Univariate Analysis

Analyze distributions of individual variables.


```python
# Histogram of Age
plt.figure(figsize=(8,5))
titanic['Age'].hist(bins=30, edgecolor='black')
plt.title('Distribution of Age')
plt.xlabel('Age')
plt.ylabel('Count')
plt.show()
```

```python
# Boxplot of Fare by Passenger Class
plt.figure(figsize=(8,5))
sns.boxplot(x='Pclass', y='Fare', data=titanic)
plt.title('Fare Distribution by Passenger Class')
plt.show()
```

**Observations:**

- Most passengers are between ages 20 and 40.
- Fare prices vary significantly by passenger class, with some outliers in fare.


## Bivariate Analysis

Explore relationships between two variables.


```python
# Survival count by Sex
plt.figure(figsize=(6,4))
sns.countplot(x='Survived', hue='Sex', data=titanic)
plt.title('Survival Count by Sex')
plt.show()
```

```python
# Survival count by Class
plt.figure(figsize=(6,4))
sns.countplot(x='Survived', hue='Pclass', data=titanic)
plt.title('Survival Count by Passenger Class')
plt.show()
```

```python
# Scatter plot Age vs Fare, colored by Survival
plt.figure(figsize=(8,6))
sns.scatterplot(x='Age', y='Fare', hue='Survived', data=titanic)
plt.title('Age vs Fare colored by Survival')
plt.show()
```

**Observations:**

- Females had higher survival rates compared to males.
- Passengers in first class had a higher survival rate.
- There is no strong visible relationship between age and fare concerning survival.


## Multivariate Analysis

Check interactions among multiple variables.


```python
# Pairplot of selected numeric columns with survival hue
sns.pairplot(titanic[['Age', 'Fare', 'Pclass', 'Survived']].dropna(), hue='Survived')
plt.show()
```

```python
# Correlation Heatmap using only numeric columns
numeric_cols = titanic.select_dtypes(include=['float64', 'int64'])
corr = numeric_cols.corr()

plt.figure(figsize=(8,6))
sns.heatmap(corr, annot=True, cmap='coolwarm')
plt.title('Correlation Heatmap')
plt.show()
```

**Observations:**

- Fare negatively correlates with passenger class (higher class = higher fare).
- Survival weakly correlates with class and fare.
- Age does not show a strong correlation with survival.
